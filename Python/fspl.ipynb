```python
import math

c = 299_792_458
f = 145.8e6
wavelength = c / f

distances_km = [300, 500, 700, 1000, 1500, 2000]

print(f"Frequency: {f/1e6:.1f} MHz")
print(f"Wavelength: {wavelength:.4f} m")
print()

for distance_km in distances_km:
    distance_m = distance_km * 1000

    fspl_db = 20 * math.log10(
        (4 * math.pi * distance_m) / wavelength
    )

    print(
        f"Distance: {distance_km:4d} km | "
        f"FSPL: {fspl_db:.2f} dB"
    )
```
