```python
import math

# -----------------------------
# Link parameters
# -----------------------------

frequency_hz = 145.8e6       # 145.8 MHz
distance_km = 1000           # satellite-ground distance
transmit_power_w = 1.0        # 1 W
transmit_gain_dbi = 3.0       # transmit antenna gain
receive_gain_dbi = 3.0        # receive antenna gain
other_losses_db = 0.0         # temporarily ignored

# -----------------------------
# Constants
# -----------------------------

c = 299_792_458               # speed of light (m/s)

# -----------------------------
# Wavelength
# -----------------------------

wavelength_m = c / frequency_hz

# -----------------------------
# Distance conversion
# -----------------------------

distance_m = distance_km * 1000

# -----------------------------
# Transmit power: W → dBm
# -----------------------------

transmit_power_dbm = 10 * math.log10(
    transmit_power_w * 1000
)

# -----------------------------
# Free-space path loss
# -----------------------------

fspl_db = 20 * math.log10(
    (4 * math.pi * distance_m) / wavelength_m
)

# -----------------------------
# Received power
# -----------------------------

received_power_dbm = (
    transmit_power_dbm
    + transmit_gain_dbi
    + receive_gain_dbi
    - fspl_db
    - other_losses_db
)

# -----------------------------
# Display
# -----------------------------

print(f"Frequency: {frequency_hz / 1e6:.1f} MHz")
print(f"Wavelength: {wavelength_m:.4f} m")
print(f"Distance: {distance_km:.0f} km")
print(f"Transmit power: {transmit_power_dbm:.2f} dBm")
print(f"Transmit antenna gain: {transmit_gain_dbi:.2f} dBi")
print(f"Receive antenna gain: {receive_gain_dbi:.2f} dBi")
print(f"Free-space path loss: {fspl_db:.2f} dB")
print(f"Other losses: {other_losses_db:.2f} dB")
print(f"Received power: {received_power_dbm:.2f} dBm")
```
